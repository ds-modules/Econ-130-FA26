## Calculating Treatment Effects

Since we have simulated a Randomized Control Trial, we can get the average causal effect by directly comparing the averages of the treatment group and the control group.

```r
treated_mean <- mean(cell_data$outcome_anxiety[cell_data$treatment_cellphone==1])
control_mean <- mean(cell_data$outcome_anxiety[cell_data$treatment_cellphone==0])

print(treated_mean)
print(control_mean)
```

```r
treatment_effect <- treated_mean - control_mean
treatment_effect
```

Let's confirm our findings using a linear regression:

```r
model <- lm(outcome_anxiety~treatment_cellphone, data = cell_data)
summary(model)
```

Note that the intercept from our linear regression matches our control mean, and that our coefficient on treatment_cellphone matches our treatment effect!

The last thing we'll do is check that our sample was balanced (that there aren't any other variables that predict prediction). We can do this by regressing treatment on other observables. If no observables predict treatment, we are more confident that the treatment was indeed random and that the control group is a good counterfactual for the treatment group.

```r
balance_test <- lm(treatment_cellphone~race+income+gender, data = cell_data)
summary(balance_test)
```

Nothing is significant so we can feel confident in our control as a counterfactual!
